# Data split
We split the training videos into training and validations

```python
from pathlib import Path
import random
import csv

dataset_path = Path("../../Nexar_DashCam_Dataset")
positive_videos = sorted((dataset_path / "train" / "positive").glob("*.mp4"))
negative_videos = sorted((dataset_path / "train" / "negative").glob("*.mp4"))
```

```python
random_generator = random.Random(42)
random_generator.shuffle(positive_videos)
random_generator.shuffle(negative_videos)
positive_split = int(len(positive_videos) * 0.8)
negative_split = int(len(negative_videos) * 0.8)
train_videos = positive_videos[:positive_split] + negative_videos[:negative_split]
validation_videos = positive_videos[positive_split:] + negative_videos[negative_split:]
print("number of training videos:", len(train_videos))
print("number of validation videos:", len(validation_videos))
```

```python
split_path = Path("video_split.csv")
with split_path.open("x", encoding="utf-8", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["video_id", "label", "split"])
    for video in train_videos:
        label = 1 if video.parent.name == "positive" else 0
        writer.writerow([video.stem, label, "train"])
    for video in validation_videos:
        label = 1 if video.parent.name == "positive" else 0
        writer.writerow([video.stem, label, "validation"])

print("save on this address:", split_path.resolve())
```
